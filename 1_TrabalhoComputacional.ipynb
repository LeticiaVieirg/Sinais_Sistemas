1. Considere o sinal de tempo contínuo definido pela seguinte expressão:
$x(t) = (t-1)\big[u(t-1) - u(t-2)\big]$

a) Plote o sinal $x(t) $ para $ 0 \leq t \leq 3$


```python
def x(t):
  y = np.zeros_like(t)
  xt = (t-1)*(np.heaviside(t-1, 1) - np.heaviside(t-2, 1))
  return xt
```

```python
t = np.linspace(0,4)
xt = x(t)

plt.figure(figsize = (10 , 4))
plt.plot(t, xt, label='x(t)', linewidth = 3)

plt.xlabel('t')
plt.ylabel('x(t)')
plt.title('Grafico: x(t)')
plt.grid(True)
plt.show()
```

b) Determine, de forma numérica, a energia do sinal $x(t) $. Compare com o resultado do item anterior.

```python
xt = x(t)
energia = np.sum(xt**2)
print(energia)
```

2. Considere o sinal $x(t)$ de tempo contínuo dado conforme a Figura 1 da lista.

a) Determine de forma analítica, uma expressão matemática para o sinal $x(t)$ para todo $t$ e plote o sinal $x(t)$.

a) Determine, de forma analıtica, uma expressão matemática para o sinal x(t) para todo t e plote o sinal x(t).

```python
def x(t):

  y=np.zeros_like(t)
  y[(t>=-1.5) & (t<0)]=2
  y[(t>=0) & (t<=3)]= 2*np.exp(-t[(t>=0) & (t<=3)]/2)

  return y
```

```python
t=np.linspace(-4,9,1000)
xt=x(t)

plt.figure(figsize=(8,3))
plt.plot(t, xt, label='x(t)', linewidth=2)

plt.xlabel('t-->')
plt.ylabel('x(t)')
plt.title('Sinal x(t)')

plt.legend()
plt.grid(True)
plt.show()
```

A expressão atende aos limites abaixo:

5. Um determinado circuito elétrico com tensão de entrada $v_i(t)$ e tensão de saída $v_o(t)$ é modelado através da seguinte equação diferencial ordinária:

$\frac{d^2v_o(t)}{dt^2} + 7\frac{dv_o(t)}{dt} + 10v_o(t) = \frac{dv_i(t)}{dt} + 6v_i(t)$

As condições iniciais para esse circuito são:

$v_o(0) = 6V$ e $\frac{dv_o(0)}{dt} = -4\frac{V}{s}$

e a entrada é $x(t) = 6e^{-3t}\,u(t)$.

Ultilizando computação simbólica, determine a expressão de saída para esse circuito.

```python
import sympy as sympy
sympy.init_printing()

t=sympy.symbols('t')
v_o=sympy.Function('v_o')(t)

eq=sympy.Eq(v_o.diff(t,t)+7*v_o.diff(t)+10*v_o,v_o.diff(t)+6*v_o)
eq
```

```python
sympy.dsolve(eq)
```

6. Um determinado sistema linear, invariante no tempo e causal é descrito pela seguinte equação diferencial:

$\frac{d^2y(t)}{dt^2} + 4\frac{dy(t)}{dt} + 3y(t) = \frac{dx(t)}{dt} + 5x(t)$
em que $y(t)$ é a saída e $x(t)$ é a entrada. Ultilizando computação simbólica, determine a resposta ao impulso para esse sistema

```python
import sympy as sp

# Variáveis simbólicas
t = sp.Symbol('t', real=True)
y = sp.Function('y')(t)

# Define delta e sua derivada
delta = sp.DiracDelta(t)
delta_p = sp.diff(delta, t)

# Equação diferencial: y'' + 4y' + 3y = δ'(t) + 5δ(t)
eq = sp.Eq(sp.diff(y, t, 2) + 4*sp.diff(y, t) + 3*y, delta_p + 5*delta)

# Resolve a equação diferencial (resposta ao impulso)
sol = sp.dsolve(eq, ics={y.subs(t, 0): 0, sp.diff(y, t).subs(t, 0): 0})

sol
```
